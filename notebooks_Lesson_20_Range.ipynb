```python
# Points projection for the remaining season
current_points    = 58
matches_remaining = 8
points_per_win    = 3

print("POINTS PROJECTION -- BEST CASE SCENARIO")
print("=" * 44)
print(f"  Current points: {current_points}")
print(f"  Matches left:   {matches_remaining}")
print()
print(f"  {'Wins':>6}  {'Total Points':>14}  {'Projection'}")
print("-" * 44)

for wins in range(0, matches_remaining + 1):
    projected = current_points + (wins * points_per_win)
    remaining_label = f"{wins} win(s) from {matches_remaining}"
    print(f"  {wins:>6}  {projected:>14}  {remaining_label}")

print("=" * 44)
```

```python
# SEASON FIXTURE AND POINTS TRACKER

print("=" * 56)
print(f"{'SEASON OVERVIEW GENERATOR':^56}")
print("=" * 56)
print()

club_name       = input("Enter club name: ")
wins_per_block  = int(input("Wins per 5 matches (0-5): "))
draws_per_block = int(input("Draws per 5 matches (0-5): "))

# Calculate losses from the inputs
losses_per_block = 5 - wins_per_block - draws_per_block

# Points per 5-match block
points_per_block = (wins_per_block * 3) + (draws_per_block * 1)

print()
print(f"  Pattern per 5 matches:")
print(f"  W{wins_per_block} D{draws_per_block} L{losses_per_block} = {points_per_block} pts")
print()
print("=" * 56)
print(f"  {'Matchday':<12} {'Points This Block':>18} {'Running Total':>18}")
print("-" * 56)

running_total = 0
total_wins    = 0
total_draws   = 0
total_losses  = 0

# A Premier League season has 38 matches -- we go in blocks of 5
# That gives us 7 complete blocks (35 matches) + 3 remaining
for block in range(1, 9):

    start_match = (block - 1) * 5 + 1
    end_match   = min(block * 5, 38)   # Don't go past match 38

    running_total  += points_per_block
    total_wins     += wins_per_block
    total_draws    += draws_per_block
    total_losses   += losses_per_block

    print(f"  MD {start_match:>2} -- {end_match:>2}    "
          f"{points_per_block:>18}    {running_total:>14} pts")

print("-" * 56)

# Final projected finish
final_points = running_total
goal_diff_estimate = (total_wins * 2) - (total_losses * 1)

print()
print(f"  PROJECTED SEASON FINISH: {club_name.upper()}")
print(f"  Projected points:        {final_points}")
print(f"  Est. goal difference:    {goal_diff_estimate:+}")
print()

if final_points >= 80:
    print("  Projection: TITLE CONTENDERS. Exceptional campaign.")
elif final_points >= 67:
    print("  Projection: CHAMPIONS LEAGUE place secured.")
elif final_points >= 55:
    print("  Projection: EUROPA LEAGUE. A respectable finish.")
elif final_points >= 40:
    print("  Projection: MID-TABLE. Safety secured, nothing more.")
else:
    print("  Projection: RELEGATION BATTLE. Urgent improvement needed.")

print()
print("=" * 56)
```
