```python
# INTERACTIVE FORMATTED MATCH REPORT

print("=" * 52)
print(f"{'MATCH REPORT GENERATOR':^52}")
print("=" * 52)
print()

# Collect match data from the user
home_team   = input("Home team: ")
away_team   = input("Away team: ")
competition = input("Competition: ")
venue       = input("Venue: ")
attendance  = int(input("Attendance: "))
home_goals  = int(input(f"Goals for {home_team}: "))
away_goals  = int(input(f"Goals for {away_team}: "))
home_shots  = int(input(f"Shots on target for {home_team}: "))
away_shots  = int(input(f"Shots on target for {away_team}: "))
home_poss   = float(input(f"Possession % for {home_team}: "))
away_poss   = 100.0 - home_poss

# Build the report
print()
print("=" * 52)
print(f"  {competition.upper()}")
print(f"  {venue}  |  Att: {attendance:,}")
print("=" * 52)
print()

# Score line centred
score_line = f"{home_team}  {home_goals} -- {away_goals}  {away_team}"
print(f"{score_line:^52}")

print()
print("-" * 52)
print(f"  {'STATISTICS':<20} {'HOME':>12} {'AWAY':>12}")
print("-" * 52)
print(f"  {'Shots on target':<20} {home_shots:>12} {away_shots:>12}")
print(f"  {'Possession':<20} {home_poss:>11.1f}% {away_poss:>11.1f}%")
print("-" * 52)
print()
print("=" * 52)
```
