```python
# ── PRE-MATCH ELIGIBILITY CHECKER ────────────────────────
# Determines if a player can feature in the next fixture

# Player details
player_name   = "Phil Foden"
club          = "Manchester City"
yellow_cards  = 4      # Accumulated in the league
red_cards     = 0      # Sent offs this season
is_injured    = False
matches_played = 28

# ── ELIGIBILITY RULES ─────────────────────────────────────
# A player is suspended if they have 5 or more yellow cards
is_suspended  = yellow_cards >= 5

# A player is available if they are not injured and not suspended
is_available  = not is_injured and not is_suspended

# A player is a key player if they have played 20 or more matches
is_key_player = matches_played >= 20

# ── REPORT ────────────────────────────────────────────────
print("=" * 48)
print("   PRE-MATCH ELIGIBILITY REPORT")
print("=" * 48)
print(f"  Player:       {player_name}")
print(f"  Club:         {club}")
print()
print(f"  Yellow cards: {yellow_cards}")
print(f"  Red cards:    {red_cards}")
print(f"  Injured:      {is_injured}")
print()
print("-" * 48)
print("  STATUS")
print("-" * 48)
print(f"  Suspended:    {is_suspended}")
print(f"  Available:    {is_available}")
print(f"  Key player:   {is_key_player}")
print()
print("=" * 48)
```
