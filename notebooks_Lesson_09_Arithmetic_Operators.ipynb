```python
# Without parentheses - Python follows the order of precedence
result_a = 3 + 4 * 2
print("3 + 4 * 2 =", result_a)    # Multiplication happens first: 3 + 8 = 11

# With parentheses - we control the order ourselves
result_b = (3 + 4) * 2
print("(3 + 4) * 2 =", result_b)  # Addition happens first: 7 * 2 = 14
```

```python
# SEASON STATISTICS CALCULATOR
# Club: Tottenham Hotspur
# Competition: Premier League

# Match results
matches_played = 34
wins   = 18
draws  = 9
losses = 7

# Goals
goals_scored   = 71
goals_conceded = 45

# Players
squad_size    = 25
unavailable   = 4

# CALCULATED STATISTICS

# Points total (win = 3 pts, draw = 1 pt)
total_points = (wins * 3) + (draws * 1)

# Goal difference
goal_difference = goals_scored - goals_conceded

# Averages
goals_per_match     = goals_scored / matches_played
conceded_per_match  = goals_conceded / matches_played

# Players available for selection
available_players = squad_size - unavailable

# Remaining matches in the season
remaining_matches = 38 - matches_played

# Points needed to reach 80 (a common European qualification target)
points_needed = 80 - total_points

# OUTPUT

print("=" * 50)
print("   SEASON REPORT -- TOTTENHAM HOTSPUR")
print("=" * 50)
print()
print(f"  Matches played:      {matches_played} of 38")
print(f"  Remaining:           {remaining_matches}")
print()
print(f"  Record:              W{wins} D{draws} L{losses}")
print(f"  Points:              {total_points}")
print(f"  Points for target:   {points_needed} more needed")
print()
print(f"  Goals scored:        {goals_scored}")
print(f"  Goals conceded:      {goals_conceded}")
print(f"  Goal difference:     {goal_difference:+}")
print()
print(f"  Goals per match:     {goals_per_match:.2f}")
print(f"  Conceded per match:  {conceded_per_match:.2f}")
print()
print(f"  Squad available:     {available_players} of {squad_size}")
print()
print("=" * 50)
```
