```python
# A player can start the match only if they are fit AND not suspended
is_fit       = True
is_suspended = False

can_start = is_fit and not is_suspended

print("Fit:           ", is_fit)
print("Suspended:     ", is_suspended)
print("Can start?     ", can_start)
```

```python
# Same player, but now they picked up an injury
is_fit       = False
is_suspended = False

can_start = is_fit and not is_suspended

print("Can start?", can_start)
```

```python
# A match is postponed if there is a storm OR the pitch is flooded
storm_warning  = False
pitch_flooded  = True

match_postponed = storm_warning or pitch_flooded

print("Storm warning: ", storm_warning)
print("Pitch flooded: ", pitch_flooded)
print("Match postponed?", match_postponed)
```

```python
# What if neither condition is present?
storm_warning  = False
pitch_flooded  = False

match_postponed = storm_warning or pitch_flooded

print("Match postponed?", match_postponed)
```

```python
# Was the player flagged offside?
offside = False

goal_stands = not offside

print("Offside flag raised:", offside)
print("Goal stands:        ", goal_stands)
```

```python
# Now the flag goes up
offside = True

goal_stands = not offside

print("Offside flag raised:", offside)
print("Goal stands:        ", goal_stands)
```

```python
# A player is selected for the squad if:
# They are fit AND (they are in form OR they are a key player)
# AND they are not serving a ban

is_fit         = True
in_form        = False
is_key_player  = True
serving_ban    = False

selected = is_fit and (in_form or is_key_player) and not serving_ban

print("Fit:            ", is_fit)
print("In form:        ", in_form)
print("Key player:     ", is_key_player)
print("Serving ban:    ", serving_ban)
print()
print("Selected for squad?", selected)
```
