```python
# 'player_name' and 'goals' are PARAMETERS -- placeholders in the definition
def print_goal_tally(player_name, goals):
    print(f"  {player_name} has scored {goals} goals this season.")

# "Mohamed Salah" and 22 are ARGUMENTS -- actual values passed in the call
print_goal_tally("Mohamed Salah", 22)
print_goal_tally("Erling Haaland", 27)
print_goal_tally("Cole Palmer", 18)
```

```python
# A player profile display function with multiple parameters
def display_player(name, club, position, goals, assists):
    contributions = goals + assists
    print(f"  {'Name':<14}: {name}")
    print(f"  {'Club':<14}: {club}")
    print(f"  {'Position':<14}: {position}")
    print(f"  {'Goals':<14}: {goals}")
    print(f"  {'Assists':<14}: {assists}")
    print(f"  {'Contributions':<14}: {contributions}")

# Calling with five arguments -- order must match the definition
print("=" * 36)
display_player("Jude Bellingham", "Real Madrid",
               "Midfielder", 23, 13)
print("=" * 36)
display_player("Bukayo Saka", "Arsenal",
               "Right Winger", 16, 11)
print("=" * 36)
```

```python
# A match report printer with a default competition name
def print_match_header(home_team, away_team,
                       competition="Premier League"):
    print("=" * 48)
    print(f"  {competition}")
    score_line = f"{home_team}  vs  {away_team}"
    print(f"  {score_line:^44}")
    print("=" * 48)

# Call without specifying competition -- uses the default
print_match_header("Arsenal", "Chelsea")

# Call with a different competition -- overrides the default
print_match_header("Arsenal", "Real Madrid",
                   "UEFA Champions League")

# Another call using the default
print_match_header("Liverpool", "Everton")
```

```python
# The same display_player function from earlier
def display_player(name, club, position, goals, assists):
    print(f"  {name} | {club} | {position} | "
          f"G:{goals} A:{assists}")

# Positional arguments -- order matters
display_player("Harry Kane", "Bayern Munich",
               "Striker", 24, 8)

# Keyword arguments -- order does not matter
display_player(
    goals     = 24,
    name      = "Harry Kane",
    assists   = 8,
    club      = "Bayern Munich",
    position  = "Striker"
)
```

```python
# A flexible fixture printer
def print_fixture(home_team, away_team, kickoff="15:00",
                  competition="Premier League",
                  venue="Home Ground"):
    print()
    print(f"  {competition}")
    print(f"  {home_team}  vs  {away_team}")
    print(f"  Kick-off: {kickoff} | Venue: {venue}")
    print()

# Using all defaults
print_fixture("Arsenal", "Chelsea")

# Overriding some defaults
print_fixture("Liverpool", "Real Madrid",
              kickoff="20:00",
              competition="Champions League",
              venue="Anfield")

# Mix of positional and keyword
print_fixture("Tottenham", "Man City",
              venue="Tottenham Hotspur Stadium",
              kickoff="12:30")
```

```python
# A function that receives a list and displays a squad roll call
def print_squad_list(squad, club_name, numbered=True):
    print(f"  {club_name.upper()} -- SQUAD LIST")
    print("-" * 40)

    for i in range(len(squad)):
        if numbered:
            print(f"  {i + 1:>2}. {squad[i]}")
        else:
            print(f"  -- {squad[i]}")

    print("-" * 40)
    print(f"  Total: {len(squad)} players")

# Calling with a list argument
city_squad = ["Ederson", "Walker", "Dias",
              "Rodri", "De Bruyne", "Haaland"]

print_squad_list(city_squad, "Manchester City")
print()
print_squad_list(city_squad, "Manchester City", numbered=False)
```

```python
# MATCH ANALYSIS SYSTEM
# A collection of functions for producing match reports

def print_report_header(home_team, away_team,
                        competition="Premier League",
                        round_info=""):
    print("=" * 54)
    print(f"  {competition.upper()}", end="")
    if round_info:
        print(f"  |  {round_info}", end="")
    print()
    score_line = f"{home_team}  vs  {away_team}"
    print(f"  {score_line:^50}")
    print("=" * 54)

def print_goal_event(minute, scorer, assist="No assist",
                     team="Unknown"):
    print(f"  {minute:>3}'  {scorer:<20}  ({team})")
    if assist != "No assist":
        print(f"        Assist: {assist}")

def print_stat_comparison(label, home_val, away_val,
                          home_label="Home", away_label="Away"):
    print(f"  {label:<22} {str(home_val):>8}   {str(away_val):>8}")

def print_verdict(home_team, home_goals,
                  away_team, away_goals):
    print()
    goal_diff = home_goals - away_goals

    if home_goals > away_goals:
        winner   = home_team
        verdict  = "WIN"
    elif home_goals == away_goals:
        winner   = None
        verdict  = "DRAW"
    else:
        winner   = away_team
        verdict  = "WIN"

    print(f"  FULL TIME: {home_team} {home_goals} -- "
          f"{away_goals} {away_team}")

    if winner:
        print(f"  RESULT: {winner} {verdict}")
        print(f"  Winning margin: {abs(goal_diff)} goal(s)")
    else:
        print(f"  RESULT: {verdict} -- Points shared")

# ── BUILDING THE REPORT ─────────────────────────

print_report_header(
    "Liverpool", "Borussia Dortmund",
    competition = "UEFA Champions League",
    round_info  = "Quarter Final"
)

print()
print("  FIRST HALF")
print("-" * 54)
print_goal_event(18, "Salah",  assist="Mac Allister",
                 team="Liverpool")
print_goal_event(37, "Hummels", team="Dortmund")

print()
print("  SECOND HALF")
print("-" * 54)
print_goal_event(55, "Nunez",  assist="Diaz",
                 team="Liverpool")
print_goal_event(83, "Salah",  assist="Alexander-Arnold",
                 team="Liverpool")

print()
print("-" * 54)
print(f"  {'STATISTICS':<22} {'Liverpool':>8}   {'Dortmund':>8}")
print("  " + "-" * 42)

print_stat_comparison("Possession",      "61%",   "39%")
print_stat_comparison("Shots",           16,      8)
print_stat_comparison("Shots on target", 7,       3)
print_stat_comparison("Corners",         9,       4)
print_stat_comparison("Fouls",           10,      13)
print_stat_comparison("Yellow cards",    1,       2)

print_verdict("Liverpool", 3, "Borussia Dortmund", 1)

print()
print("=" * 54)
```
