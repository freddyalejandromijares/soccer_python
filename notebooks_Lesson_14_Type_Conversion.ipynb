```python
# The four data types you know
player_name   = "Bukayo Saka"    # str   -- text
squad_number  = 7                # int   -- whole number
pass_accuracy = 87.4             # float -- decimal number
is_available  = True             # bool  -- True or False

print(type(player_name))
print(type(squad_number))
print(type(pass_accuracy))
print(type(is_available))
```

```python
# Imagine these two values came from user input
# input() always returns a string, remember?
goals_a = "10"     # This looks like a number but it is a string
goals_b = "20"     # This one too

total = goals_a + goals_b

print("Total goals:", total)
```

```python
# Converting a string to an integer
goals_string = "23"
goals_int    = int(goals_string)

print(goals_int)
print(type(goals_int))
```

```python
# The correct version -- converting before adding
goals_a = "10"
goals_b = "20"

total = int(goals_a) + int(goals_b)

print("Total goals:", total)
```

```python
# Float to integer -- the decimal is cut off, not rounded
average_goals = 2.9

whole_number = int(average_goals)

print("Converted value:", whole_number)
```

```python
# Converting an integer to a float
goals   = 15
matches = 22

# Integer division in older code could give a rounded result
# Using float() ensures we get the full decimal answer
goals_float   = float(goals)
average       = goals_float / matches

print(f"Goals per match: {average:.2f}")
```

```python
# Converting a string decimal to a real float
possession_string = "63.5"
possession_float  = float(possession_string)

adjusted = possession_float - 10.0

print(f"Adjusted possession: {adjusted:.1f}%")
```

```python
# This causes an error -- you cannot concatenate a string and an integer
goals = 25
# message = "Goals scored: " + goals   # This would crash!

# str() fixes it
message = "Goals scored: " + str(goals)
print(message)
```

```python
# Collecting and converting different types of input
print("=== PLAYER DATA ENTRY ===")
print()

player_name   = input("Player name: ")
squad_number  = int(input("Squad number: "))
age           = int(input("Age: "))
goals         = int(input("Goals this season: "))
matches       = int(input("Matches played: "))
pass_accuracy = float(input("Pass accuracy (%): "))

print()
print("Data received and converted successfully!")
print()

# Now we can do real calculations because the types are correct
goals_per_match = goals / matches
remaining_years = 35 - age    # Estimated years until retirement

print(f"Goals per match:    {goals_per_match:.2f}")
print(f"Est. years left:    {remaining_years}")
print(f"Pass accuracy:      {pass_accuracy:.1f}%")
```

```python
# SEASON STATS CONVERTER AND REPORT GENERATOR

print("=" * 50)
print(f"{'SEASON STATS REPORT':^50}")
print("=" * 50)
print()
print("Enter the season data below.")
print()

# Collecting data
club_name      = input("Club name: ")
matches_played = int(input("Matches played: "))
wins           = int(input("Wins: "))
draws          = int(input("Draws: "))
losses         = int(input("Losses: "))
goals_for      = int(input("Goals scored: "))
goals_against  = int(input("Goals conceded: "))
squad_budget   = float(input("Transfer budget spent (millions): "))

# Type check before calculations
print()
print("Converting data...")
print(f"  matches_played is a:  {type(matches_played)}")
print(f"  squad_budget is a:    {type(squad_budget)}")
print()

# Calculations -- all safe because types are correct
points          = (wins * 3) + (draws * 1)
goal_difference = goals_for - goals_against
goals_per_match = goals_for / matches_played
conceded_per_match = goals_against / matches_played
remaining       = 38 - matches_played

# Output report
print("=" * 50)
print(f"  {'FINAL REPORT':^46}")
print("=" * 50)
print(f"  Club:              {club_name}")
print(f"  Played:            {matches_played} of 38")
print(f"  Remaining:         {remaining}")
print()
print(f"  Wins:              {wins}")
print(f"  Draws:             {draws}")
print(f"  Losses:            {losses}")
print(f"  Points:            {points}")
print()
print(f"  Goals scored:      {goals_for}")
print(f"  Goals conceded:    {goals_against}")
print(f"  Goal difference:   {goal_difference:+}")
print()
print(f"  Goals per match:   {goals_per_match:.2f}")
print(f"  Conceded p/match:  {conceded_per_match:.2f}")
print()
print(f"  Budget spent:      £{squad_budget:,.1f}m")
print("=" * 50)
```
