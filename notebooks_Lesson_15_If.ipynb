```python
# Goals-scored checker with user input
print("=== GOAL TALLY CHECKER ===")
print()

player_name = input("Player name: ")
goals       = int(input("Goals this season: "))

print()

if goals >= 20:
    print(f"Outstanding season for {player_name}!")
    print(f"{goals} goals is a genuine Golden Boot contender.")

if goals >= 10:
    print(f"{player_name} has reached double figures. Solid contribution.")

if goals == 0:
    print(f"No goals yet for {player_name}. The manager will want more.")

print()
print("Analysis complete.")
```

```python
# PRE-MATCH STATUS REPORT
# Uses if statements to flag important squad information

print("=" * 50)
print(f"{'PRE-MATCH STATUS REPORT':^50}")
print("=" * 50)
print()

# Collect squad data from the user
club_name      = input("Club name: ")
players_fit    = int(input("Fit players available: "))
yellow_cards   = int(input("Yellow cards (squad total): "))
goals_scored   = int(input("Goals scored last match: "))
goals_conceded = int(input("Goals conceded last match: "))
days_since_rest = int(input("Days since last match: "))

print()
print("=" * 50)
print(f"  {club_name.upper()} -- MATCH DAY STATUS")
print("=" * 50)
print()

# Squad availability check
if players_fit < 14:
    print("  SQUAD ALERT: Availability is critically low.")
    print(f"  Only {players_fit} players fit. Manager must act.")
    print()

# Disciplinary check
if yellow_cards >= 10:
    print("  DISCIPLINE WARNING: High card count across the squad.")
    print(f"  {yellow_cards} yellows accumulated. Risk of suspensions.")
    print()

# Recent form check
if goals_scored >= 3:
    print("  ATTACK: Excellent form in front of goal last match.")
    print(f"  {goals_scored} goals scored. Keep that momentum going.")
    print()

if goals_conceded >= 3:
    print("  DEFENCE: Defensive issues flagged from last match.")
    print(f"  {goals_conceded} goals conceded. The back line needs work.")
    print()

# Fatigue check
if days_since_rest <= 3:
    print("  FATIGUE: Short turnaround since the last fixture.")
    print(f"  Only {days_since_rest} days of recovery. Rotation advised.")
    print()

# All clear message
if players_fit >= 18 and yellow_cards < 10 and days_since_rest >= 4:
    print("  ALL CLEAR: No major concerns flagged for this fixture.")
    print()

print("=" * 50)
print("  End of pre-match report.")
print("=" * 50)
```
