```python
# Season statistics — Liverpool FC
wins = 22
draws = 8
losses = 8

points_per_win  = 3
points_per_draw = 1

# Calculate total points
total_points = (wins * points_per_win) + (draws * points_per_draw)

# Calculate goals and goal difference
goals_scored   = 84
goals_conceded = 41
goal_difference = goals_scored - goals_conceded

# Display the season summary
print("=" * 42)
print("   SEASON SUMMARY — LIVERPOOL FC")
print("=" * 42)
print("Played:", wins + draws + losses)
print("Won:   ", wins)
print("Drawn: ", draws)
print("Lost:  ", losses)
print()
print("Goals scored:    ", goals_scored)
print("Goals conceded:  ", goals_conceded)
print("Goal difference: ", goal_difference)
print()
print("TOTAL POINTS:    ", total_points)
print("=" * 42)
```
