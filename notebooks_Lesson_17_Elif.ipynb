```python
# PLAYER SEASON RATING SYSTEM
# Produces a performance verdict based on season statistics

print("=" * 54)
print(f"{'PLAYER SEASON RATING SYSTEM':^54}")
print("=" * 54)
print()

# Collect player data
player_name    = input("Player name: ")
position       = input("Position (Forward/Midfielder/Defender): ")
goals          = int(input("Goals this season: "))
assists        = int(input("Assists this season: "))
matches        = int(input("Matches played: "))
yellow_cards   = int(input("Yellow cards: "))
rating_score   = float(input("Average match rating (1.0 to 10.0): "))

# Calculations
contributions  = goals + assists
goals_per_match = goals / matches

print()
print("=" * 54)
print(f"  {player_name.upper()}")
print(f"  {position} | {matches} appearances")
print("=" * 54)
print()

# Goals verdict
print("  GOALS:")
if goals >= 20:
    print(f"  Outstanding. {goals} goals is a top-class return.")
elif goals >= 12:
    print(f"  Very good. {goals} goals shows real quality.")
elif goals >= 6:
    print(f"  Decent. {goals} goals but room for improvement.")
elif goals >= 1:
    print(f"  Modest. Only {goals} goals from {matches} appearances.")
else:
    print("  No goals registered. A difficult season in front of goal.")

print()

# Contributions verdict
print("  OVERALL CONTRIBUTION:")
if contributions >= 30:
    print(f"  Elite level. {contributions} goal contributions is world-class.")
elif contributions >= 20:
    print(f"  Excellent. {contributions} contributions. A key player all season.")
elif contributions >= 10:
    print(f"  Solid. {contributions} contributions from {matches} matches.")
else:
    print(f"  Limited. {contributions} contributions. Consistency is needed.")

print()

# Discipline verdict
print("  DISCIPLINE:")
if yellow_cards == 0:
    print("  Perfect discipline. Zero bookings all season.")
elif yellow_cards <= 3:
    print(f"  Good. Only {yellow_cards} yellow cards. No major concerns.")
elif yellow_cards <= 6:
    print(f"  Caution. {yellow_cards} yellows is edging towards problematic.")
else:
    print(f"  Poor discipline. {yellow_cards} yellow cards is a real concern.")

print()

# Overall rating verdict
print("  SEASON RATING:")
if rating_score >= 8.0:
    print(f"  WORLD-CLASS season. Average rating of {rating_score:.1f}.")
    print("  One of the best players in the league this year.")
elif rating_score >= 7.0:
    print(f"  EXCELLENT season. Average rating of {rating_score:.1f}.")
    print("  Consistent, impactful and a joy to watch.")
elif rating_score >= 6.0:
    print(f"  GOOD season. Average rating of {rating_score:.1f}.")
    print("  Contributed well but the ceiling is higher.")
elif rating_score >= 5.0:
    print(f"  AVERAGE season. Average rating of {rating_score:.1f}.")
    print("  Some good moments but too inconsistent overall.")
else:
    print(f"  POOR season. Average rating of {rating_score:.1f}.")
    print("  A difficult year. The manager will expect more next season.")

print()
print("=" * 54)
print(f"  Goals per match:  {goals_per_match:.2f}")
print(f"  Contributions:    {contributions}")
print("=" * 54)
```
