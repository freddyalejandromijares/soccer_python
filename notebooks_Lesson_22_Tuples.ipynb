```python
# A list uses square brackets -- mutable
squad_list  = ["Alisson", "Van Dijk", "Salah"]

# A tuple uses parentheses -- immutable
squad_tuple = ("Alisson", "Van Dijk", "Salah")

print(squad_list)
print(squad_tuple)
print(type(squad_list))
print(type(squad_tuple))
```

```python
# Club founding information -- this never changes
club_info = ("Manchester City", 1880, "City of Manchester Stadium", "Sky Blues")

# Official pitch dimensions -- fixed by the rules of football
pitch_dimensions = (105, 68)    # Length x Width in metres

# A historic match result -- permanently recorded
historic_result = ("Real Madrid", 14, "Malmö FF", 0, 1960)

# A player's date of birth -- immutable personal fact
erling_dob = (21, 7, 2000)   # Day, Month, Year

print(club_info)
print(pitch_dimensions)
print(historic_result)
print(erling_dob)
```

```python
# Club info tuple
club_info = ("Manchester City", 1880, "City of Manchester Stadium", "Sky Blues")

# Accessing individual items by index
print(f"Club name:    {club_info[0]}")
print(f"Founded:      {club_info[1]}")
print(f"Stadium:      {club_info[2]}")
print(f"Nickname:     {club_info[3]}")
print()

# Negative indexing works too
print(f"Last item:    {club_info[-1]}")
print(f"Second last:  {club_info[-2]}")
```

```python
# Attempting to change a tuple -- this will cause a TypeError
pitch_dimensions = (105, 68)

pitch_dimensions[0] = 110    # This line will cause an error!
```

```python
# Tuple unpacking -- one line assigns all values at once
match_result = ("Arsenal", 3, "Chelsea", 1)

home_team, home_goals, away_team, away_goals = match_result

print(f"Home: {home_team} {home_goals}")
print(f"Away: {away_team} {away_goals}")
```

```python
# A tuple of Champions League winning clubs (recent years)
cl_winners = (
    "Real Madrid",
    "Manchester City",
    "Real Madrid",
    "Chelsea",
    "Bayern Munich",
    "Liverpool"
)

print("RECENT CHAMPIONS LEAGUE WINNERS")
print("=" * 36)

for i, club in enumerate(cl_winners):
    year = 2024 - i
    print(f"  {year}: {club}")

print("=" * 36)
print(f"  Total entries: {len(cl_winners)}")
```

```python
# A list of tuples -- each tuple is one match result
# Format: (home_team, home_goals, away_team, away_goals)
results = [
    ("Arsenal",          3, "Chelsea",         1),
    ("Liverpool",        2, "Manchester City",  2),
    ("Tottenham",        0, "Aston Villa",      1),
    ("Newcastle",        4, "Wolves",           0),
    ("Manchester Utd",   1, "Everton",          1),
]

print("PREMIER LEAGUE RESULTS")
print("=" * 50)
print(f"  {'Home':<20} {'Score':^7} {'Away':<20}")
print("-" * 50)

for home, hg, away, ag in results:
    score = f"{hg} -- {ag}"
    print(f"  {home:<20} {score:^7} {away:<20}")

print("=" * 50)
print(f"  Matches shown: {len(results)}")
```

```python
# CLUB HISTORY REPORT
# Uses tuples to store permanent and historical club data

print("=" * 54)
print(f"{'CLUB HISTORY REPORT':^54}")
print("=" * 54)
print()

# Club information tuple -- permanent facts
club_info = (
    "Liverpool FC",
    1892,
    "Anfield",
    54074,
    "The Reds",
    "You'll Never Walk Alone"
)

name, founded, stadium, capacity, nickname, anthem = club_info

print(f"  Club:      {name}")
print(f"  Founded:   {founded}")
print(f"  Stadium:   {stadium}")
print(f"  Capacity:  {capacity:,}")
print(f"  Nickname:  {nickname}")
print(f"  Anthem:    {anthem}")
print()

# Trophy history -- list of tuples (competition, year, opponent, score)
trophies = [
    ("Champions League", 2019, "Tottenham",     "2-0"),
    ("Premier League",   2020, "Season title",  "99 pts"),
    ("FA Cup",           2022, "Chelsea",       "6-5 pens"),
    ("League Cup",       2022, "Chelsea",       "11-10 pens"),
    ("Champions League", 2005, "AC Milan",      "3-3 (3-2 pens)"),
    ("Champions League", 2001, "Alaves",        "5-4"),
]

print("-" * 54)
print("  TROPHY CABINET")
print("-" * 54)
print(f"  {'Competition':<22} {'Year':>5}  {'Opponent/Detail'}")
print("-" * 54)

for competition, year, opponent, score in trophies:
    print(f"  {competition:<22} {year:>5}  {opponent} ({score})")

print("-" * 54)
print(f"  Trophies shown: {len(trophies)}")
print()

# Historic match -- a single permanent tuple
historic_match = ("Liverpool", 3, "AC Milan", 3, "Istanbul", 2005)
hclub, hg, aclub, ag, venue, hist_year = historic_match

print("=" * 54)
print("  MOST MEMORABLE MATCH")
print("=" * 54)
print(f"  {hclub} {hg} -- {ag} {aclub}")
print(f"  Venue: {venue} | Year: {hist_year}")
print(f"  (Liverpool won on penalties)")
print("=" * 54)
```
