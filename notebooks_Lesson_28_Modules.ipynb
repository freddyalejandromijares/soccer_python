```python
import random

# RANDOM MATCH SIMULATOR

def simulate_match(home_team, away_team,
                   home_advantage=True):

    # Home teams score slightly more on average
    home_max = 5 if home_advantage else 4
    away_max = 4

    home_goals = random.randint(0, home_max)
    away_goals = random.randint(0, away_max)

    return home_goals, away_goals

def generate_scorers(team_name, num_goals, squad):
    events = []
    for _ in range(num_goals):
        scorer = random.choice(squad)
        minute = random.randint(1, 90)
        events.append((minute, scorer, team_name))
    events.sort(key=lambda x: x[0])
    return events

def print_match_report(home_team, away_team,
                       home_goals, away_goals,
                       all_events):

    print("=" * 52)
    print(f"  MATCH SIMULATION")
    print("=" * 52)
    print(f"  {home_team:<22} {home_goals} -- {away_goals}  {away_team}")
    print("=" * 52)

    if all_events:
        print()
        print("  GOALS:")
        print("-" * 52)
        for minute, scorer, team in all_events:
            print(f"  {minute:>3}'  {scorer:<20}  ({team})")
    else:
        print()
        print("  No goals scored.")

    print()

    if home_goals > away_goals:
        print(f"  RESULT: {home_team} WIN")
    elif home_goals == away_goals:
        print("  RESULT: DRAW")
    else:
        print(f"  RESULT: {away_team} WIN")

    print("=" * 52)

# Squad lists
liverpool_squad  = ["Alisson", "Van Dijk", "Salah",
                    "Nunez", "Diaz", "Mac Allister",
                    "Szoboszlai", "Robertson", "Elliott"]

arsenal_squad    = ["Raya", "White", "Saliba",
                    "Saka", "Havertz", "Odegaard",
                    "Martinelli", "Rice", "Trossard"]

# Run the simulation
print("Running simulation...")
print()

home_goals, away_goals = simulate_match(
    "Liverpool", "Arsenal", home_advantage=True
)

home_events = generate_scorers(
    "Liverpool", home_goals, liverpool_squad
)
away_events = generate_scorers(
    "Arsenal", away_goals, arsenal_squad
)

all_events = home_events + away_events
all_events.sort(key=lambda x: x[0])

print_match_report(
    "Liverpool", "Arsenal",
    home_goals, away_goals,
    all_events
)
```
