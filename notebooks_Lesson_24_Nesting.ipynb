```python
# Squad database -- a list of dictionaries
# Each dictionary is one complete player profile
squad = [
    {
        "name":     "Alisson Becker",
        "position": "GK",
        "age":      31,
        "goals":    0,
        "assists":  1,
        "rating":   8.1
    },
    {
        "name":     "Virgil van Dijk",
        "position": "CB",
        "age":      32,
        "goals":    3,
        "assists":  2,
        "rating":   8.4
    },
    {
        "name":     "Mohamed Salah",
        "position": "RW",
        "age":      31,
        "goals":    22,
        "assists":  13,
        "rating":   8.7
    },
    {
        "name":     "Darwin Nunez",
        "position": "ST",
        "age":      24,
        "goals":    17,
        "assists":  8,
        "rating":   7.6
    },
    {
        "name":     "Alexis Mac Allister",
        "position": "CM",
        "age":      25,
        "goals":    7,
        "assists":  9,
        "rating":   8.0
    }
]

print(f"Squad loaded: {len(squad)} players")
```

```python
# Access the first player -- index 0
first_player = squad[0]
print(f"First player: {first_player['name']}")

# Access in one line: list index then dictionary key
print(f"Third player's goals: {squad[2]['goals']}")
print(f"Last player's rating: {squad[-1]['rating']}")
```

```python
# Find players who are eligible for a new long-term contract
# Criteria: age under 28 AND rating above 7.8

print("CONTRACT RENEWAL CANDIDATES")
print("=" * 44)
print("  (Age under 28 AND Rating above 7.8)")
print("-" * 44)

candidates_found = 0

for player in squad:
    if player["age"] < 28 and player["rating"] > 7.8:
        print(f"  {player['name']} "
              f"(Age: {player['age']}, "
              f"Rating: {player['rating']})")
        candidates_found += 1

print("-" * 44)
print(f"  Candidates found: {candidates_found}")
print("=" * 44)
```

```python
# Season statistics summary from the squad database
total_goals   = 0
total_assists = 0
total_rating  = 0
top_scorer    = squad[0]

for player in squad:
    total_goals   += player["goals"]
    total_assists += player["assists"]
    total_rating  += player["rating"]

    if player["goals"] > top_scorer["goals"]:
        top_scorer = player

average_rating = total_rating / len(squad)

print("SQUAD SEASON SUMMARY")
print("=" * 38)
print(f"  Total goals:      {total_goals}")
print(f"  Total assists:    {total_assists}")
print(f"  Average rating:   {average_rating:.2f}")
print(f"  Top scorer:       {top_scorer['name']} "
      f"({top_scorer['goals']} goals)")
print("=" * 38)
```

```python
# INTERACTIVE SQUAD DATABASE AND REPORT
# Builds a list of player dictionaries from user input

print("=" * 56)
print(f"{'SQUAD DATABASE BUILDER':^56}")
print("=" * 56)
print()

club_name   = input("Club name: ")
squad_size  = int(input("How many players to enter (max 5): "))
print()

# Build the squad database from user input
squad = []

for i in range(squad_size):
    print(f"  --- Player {i + 1} of {squad_size} ---")

    name     = input("  Name: ")
    position = input("  Position: ")
    age      = int(input("  Age: "))
    goals    = int(input("  Goals: "))
    assists  = int(input("  Assists: "))
    rating   = float(input("  Rating (1-10): "))

    player = {
        "name":          name,
        "position":      position,
        "age":           age,
        "goals":         goals,
        "assists":       assists,
        "contributions": goals + assists,
        "rating":        rating
    }

    squad.append(player)
    print(f"  {name} added to the squad.")
    print()

# Squad report header
print("=" * 56)
print(f"  {club_name.upper()} -- FULL SQUAD REPORT")
print("=" * 56)
print(f"  {'Name':<22} {'Pos':<5} {'Age':>4} "
      f"{'G':>4} {'A':>4} {'Contrib':>8} {'Rating':>7}")
print("-" * 56)

# Accumulators
total_goals       = 0
total_assists     = 0
total_rating      = 0
top_scorer        = squad[0]
best_rated        = squad[0]

for player in squad:
    print(f"  {player['name']:<22} "
          f"{player['position']:<5} "
          f"{player['age']:>4} "
          f"{player['goals']:>4} "
          f"{player['assists']:>4} "
          f"{player['contributions']:>8} "
          f"{player['rating']:>7.1f}")

    total_goals   += player["goals"]
    total_assists += player["assists"]
    total_rating  += player["rating"]

    if player["goals"] > top_scorer["goals"]:
        top_scorer = player

    if player["rating"] > best_rated["rating"]:
        best_rated = player

# Summary section
average_rating = total_rating / len(squad)

print("=" * 56)
print()
print("  SQUAD SUMMARY")
print("-" * 56)
print(f"  Players in database:  {len(squad)}")
print(f"  Total goals:          {total_goals}")
print(f"  Total assists:        {total_assists}")
print(f"  Total contributions:  {total_goals + total_assists}")
print(f"  Average rating:       {average_rating:.2f}")
print()
print(f"  Top scorer:    {top_scorer['name']} "
      f"({top_scorer['goals']} goals)")
print(f"  Best rated:    {best_rated['name']} "
      f"({best_rated['rating']} avg rating)")

# Young talent check
print()
print("  YOUNG TALENTS (under 24):")
young_found = False

for player in squad:
    if player["age"] < 24:
        print(f"  -- {player['name']} "
              f"(age {player['age']}, "
              f"rating {player['rating']})")
        young_found = True

if not young_found:
    print("  No players under 24 in this database.")

print()
print("=" * 56)
```
