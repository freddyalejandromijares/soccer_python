```python
# Without a function -- repeated logic
print("=" * 44)
print(f"  {'Erling Haaland':^40}")
print(f"  {'Manchester City':^40}")
print("=" * 44)

# ... some programme code in between ...

print("=" * 44)
print(f"  {'Mohamed Salah':^40}")
print(f"  {'Liverpool':^40}")
print("=" * 44)
```

```python
# With a function -- define once, use anywhere
def print_player_header(name, club):
    print("=" * 44)
    print(f"  {name:^40}")
    print(f"  {club:^40}")
    print("=" * 44)

# Now you just call it
print_player_header("Erling Haaland", "Manchester City")
print_player_header("Mohamed Salah", "Liverpool")
print_player_header("Jude Bellingham", "Real Madrid")
```

```python
# A function that prints the match day banner
def print_match_banner():
    print()
    print("*" * 50)
    print(f"{'MATCH DAY':^50}")
    print(f"{'Premier League':^50}")
    print("*" * 50)
    print()

# Call it whenever a match report begins
print_match_banner()
print("  Arsenal vs Chelsea")
print("  Emirates Stadium | 17:30 KO")
print_match_banner()
print("  Liverpool vs Manchester City")
print("  Anfield | 20:00 KO")
```

```python
# A function that classifies a match result
def classify_result(home_goals, away_goals):
    if home_goals > away_goals:
        result = "HOME WIN"
    elif home_goals == away_goals:
        result = "DRAW"
    else:
        result = "AWAY WIN"
    print(f"  Result: {result}")

# Using the function for different matches
classify_result(3, 1)
classify_result(2, 2)
classify_result(0, 1)
```

```python
# A squad fitness checker function
def check_fitness(player_name, fitness_score):
    print(f"  {player_name:<22}", end="")

    if fitness_score >= 85:
        print(f"  {fitness_score}  -- FULLY FIT. Start.")
    elif fitness_score >= 70:
        print(f"  {fitness_score}  -- FIT. Available.")
    elif fitness_score >= 55:
        print(f"  {fitness_score}  -- LIMITED. Bench only.")
    else:
        print(f"  {fitness_score}  -- UNFIT. Rest required.")

# Pre-match fitness report
print("PRE-MATCH FITNESS REPORT")
print("=" * 52)
print(f"  {'Player':<22}  {'Score'}  {'Status'}")
print("-" * 52)

check_fitness("Ederson",         92)
check_fitness("Ruben Dias",      88)
check_fitness("Rodri",           61)
check_fitness("Kevin De Bruyne", 45)
check_fitness("Erling Haaland",  95)

print("=" * 52)
```

```python
# MATCH REPORT GENERATOR -- USING FUNCTIONS

# Function 1: Print a formatted section header
def print_section(title):
    print()
    print("-" * 50)
    print(f"  {title.upper()}")
    print("-" * 50)

# Function 2: Print a match title block
def print_match_title(home, away, competition):
    print("=" * 50)
    print(f"  {competition.upper()}")
    score_line = f"{home}  vs  {away}"
    print(f"  {score_line:^46}")
    print("=" * 50)

# Function 3: Display a goal event
def print_goal(minute, scorer, assist, team):
    print(f"  {minute:>3}' GOAL -- {scorer} ({team})")
    print(f"       Assisted by: {assist}")

# Function 4: Print a stat comparison row
def print_stat(label, home_value, away_value):
    print(f"  {label:<20} {str(home_value):>8}  {str(away_value):>8}")

# Function 5: Classify and print the final result
def print_result(home_team, home_goals, away_team, away_goals):
    print()
    if home_goals > away_goals:
        result = f"{home_team} WIN"
    elif home_goals == away_goals:
        result = "DRAW"
    else:
        result = f"{away_team} WIN"

    print(f"  FULL TIME: {home_team} {home_goals} -- "
          f"{away_goals} {away_team}")
    print(f"  RESULT: {result}")

# ── MATCH DATA ───────────────────────────────────
home_team   = "Arsenal"
away_team   = "Real Madrid"
competition = "UEFA Champions League"
venue       = "Emirates Stadium"
attendance  = 60350

home_goals_total = 2
away_goals_total = 1

# ── BUILD THE REPORT USING FUNCTIONS ─────────────

print_match_title(home_team, away_team, competition)

print(f"  Venue:      {venue}")
print(f"  Attendance: {attendance:,}")

print_section("First Half Goals")
print_goal(23, "Saka",     "Odegaard",  home_team)
print_goal(41, "Vinicius", "Bellingham", away_team)

print_section("Second Half Goals")
print_goal(78, "Havertz",  "Martinelli", home_team)

print_section("Match Statistics")
print(f"  {'Stat':<20} {'Arsenal':>8}  {'Real Mad':>8}")
print("  " + "-" * 40)
print_stat("Possession",      "58%",    "42%")
print_stat("Shots",           14,       9)
print_stat("Shots on target", 6,        3)
print_stat("Corners",         7,        4)
print_stat("Yellow cards",    1,        2)

print_section("Full Time")
print_result(home_team, home_goals_total, away_team, away_goals_total)

print()
print("=" * 50)
```
