```python
# Flagging players who need a contract review
# Any player under 75 overall needs attention

player_ratings = [
    ("Ederson",          88),
    ("Josko Gvardiol",   82),
    ("Rodri",            91),
    ("Kevin De Bruyne",  91),
    ("Phil Foden",       88),
    ("Erling Haaland",   92),
    ("Jeremy Doku",      81),
]

print("CONTRACT REVIEW -- MANCHESTER CITY")
print("=" * 40)

for player, rating in player_ratings:
    if rating >= 85:
        status = "ELITE -- No changes needed"
    elif rating >= 80:
        status = "GOOD -- Review in 12 months"
    else:
        status = "REVIEW NEEDED -- Negotiate now"

    print(f"  {player:<22} {rating}  {status}")

print("=" * 40)
```

```python
# Calculate season statistics from a list of match results
match_goals = [2, 1, 3, 0, 2, 4, 1, 2, 3, 0,
               1, 2, 3, 2, 1, 4, 2, 1, 3, 2]

total_goals    = 0
clean_sheets   = 0
big_wins       = 0   # matches with 3 or more goals

for goals in match_goals:
    total_goals += goals

    if goals == 0:
        clean_sheets += 1

    if goals >= 3:
        big_wins += 1

matches = len(match_goals)
average = total_goals / matches

print("=" * 42)
print("  SEASON GOALS ANALYSIS")
print("=" * 42)
print(f"  Matches played:    {matches}")
print(f"  Total goals:       {total_goals}")
print(f"  Goals per match:   {average:.2f}")
print(f"  Clean sheets:      {clean_sheets}")
print(f"  High-scoring games: {big_wins}")
print("=" * 42)
```

```python
# SQUAD REPORT GENERATOR
# Produces a full assessment of a squad based on player ratings

print("=" * 54)
print(f"{'SQUAD REPORT GENERATOR':^54}")
print("=" * 54)
print()

club_name = input("Enter club name: ")
print()

# Squad data -- each entry is (name, position, overall_rating)
squad = [
    ("Alisson Becker",        "GK",  89),
    ("Trent Alexander-Arnold","RB",  87),
    ("Virgil van Dijk",       "CB",  90),
    ("Andrew Robertson",      "LB",  85),
    ("Alexis Mac Allister",   "CM",  84),
    ("Dominik Szoboszlai",    "CM",  84),
    ("Mohamed Salah",         "RW",  89),
    ("Darwin Nunez",          "ST",  84),
    ("Luis Diaz",             "LW",  85),
    ("Curtis Jones",          "CM",  79),
    ("Harvey Elliott",        "RW",  78),
]

# Accumulators
total_rating   = 0
elite_count    = 0
concern_count  = 0

print(f"  CLUB: {club_name.upper()}")
print(f"  SQUAD SIZE: {len(squad)} players")
print()
print(f"  {'Name':<26} {'Pos':<5} {'OVR':>4}  {'Status'}")
print("-" * 54)

for name, position, rating in squad:

    total_rating += rating

    if rating >= 88:
        status = "World Class"
        elite_count += 1
    elif rating >= 83:
        status = "Very Good"
    elif rating >= 78:
        status = "Good"
    else:
        status = "Needs improvement"
        concern_count += 1

    print(f"  {name:<26} {position:<5} {rating:>4}  {status}")

# Summary
average_rating = total_rating / len(squad)

print("-" * 54)
print()
print(f"  Average squad rating:  {average_rating:.1f}")
print(f"  World Class players:   {elite_count}")
print(f"  Players of concern:    {concern_count}")
print()
print("=" * 54)
```
