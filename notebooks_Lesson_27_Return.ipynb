```python
# A simple function that returns a calculated value
def calculate_points(wins, draws):
    points = (wins * 3) + (draws * 1)
    return points

# The returned value is caught and stored in a variable
arsenal_points  = calculate_points(20, 7)
city_points     = calculate_points(19, 8)
liverpool_points = calculate_points(18, 9)

print(f"Arsenal points:   {arsenal_points}")
print(f"City points:      {city_points}")
print(f"Liverpool points: {liverpool_points}")
```

```python
# Version 1: Using print() inside the function
def goals_per_match_print(goals, matches):
    result = goals / matches
    print(f"Goals per match: {result:.2f}")

# Version 2: Using return
def goals_per_match_return(goals, matches):
    result = goals / matches
    return result

# Calling the print version
goals_per_match_print(22, 35)
# We cannot do anything with the result -- it disappears after display

# Calling the return version
average = goals_per_match_return(22, 35)
# Now we can use the result
print(f"Goals per match: {average:.2f}")
print(f"Double the average: {average * 2:.2f}")
print(f"Above 0.5 target: {average > 0.5}")
```

```python
def check_goal_line(ball_position):
    if ball_position > 100:
        return "GOAL -- The ball crossed the line!"

    return "No goal -- Ball did not cross the line."

    print("This line never runs.")    # Unreachable code

result = check_goal_line(105)
print(result)

result = check_goal_line(98)
print(result)
```

```python
# Returns a classification string based on the overall rating
def classify_player(overall_rating):
    if overall_rating >= 90:
        return "World Class"
    elif overall_rating >= 85:
        return "Elite"
    elif overall_rating >= 78:
        return "Very Good"
    elif overall_rating >= 70:
        return "Good"
    else:
        return "Development Player"

# Using the returned values
players = [
    ("Erling Haaland",   92),
    ("Virgil van Dijk",  90),
    ("Declan Rice",      85),
    ("Phil Foden",       88),
    ("Curtis Jones",     79),
    ("Ben Doak",         71),
]

print("PLAYER CLASSIFICATION REPORT")
print("=" * 44)
print(f"  {'Player':<22} {'OVR':>4}  {'Class'}")
print("-" * 44)

for name, rating in players:
    classification = classify_player(rating)
    print(f"  {name:<22} {rating:>4}  {classification}")

print("=" * 44)
```

```python
# Returns multiple calculated stats at once
def analyse_season(wins, draws, losses, goals_for, goals_against):
    matches_played = wins + draws + losses
    points         = (wins * 3) + (draws * 1)
    goal_difference = goals_for - goals_against
    goals_per_match = goals_for / matches_played

    return points, goal_difference, goals_per_match

# Unpacking the returned tuple into separate variables
pts, gd, avg = analyse_season(20, 7, 11, 72, 48)

print(f"Points:          {pts}")
print(f"Goal difference: {gd:+}")
print(f"Goals per match: {avg:.2f}")
```

```python
def calculate_points(wins, draws):
    return (wins * 3) + (draws * 1)

def classify_season(points):
    if points >= 80:
        return "Title winning"
    elif points >= 67:
        return "Champions League"
    elif points >= 55:
        return "Europa League"
    else:
        return "Mid table or below"

# Using the return value directly in print()
print(f"Points: {calculate_points(22, 6)}")

# Passing one function's return directly into another
season_type = classify_season(calculate_points(22, 6))
print(f"Season type: {season_type}")
```

```python
# PLAYER RATING ENGINE
# A system of functions that calculate, classify and report

def calculate_contributions(goals, assists):
    return goals + assists

def calculate_goals_per_match(goals, matches):
    if matches == 0:
        return 0.0
    return goals / matches

def get_performance_grade(rating):
    if rating >= 8.5:
        return "S", "Outstanding"
    elif rating >= 7.5:
        return "A", "Excellent"
    elif rating >= 6.5:
        return "B", "Good"
    elif rating >= 5.5:
        return "C", "Average"
    else:
        return "D", "Below Standard"

def get_transfer_value(overall, age):
    base_value = (overall - 60) * 2.5
    if age < 23:
        age_multiplier = 1.8
    elif age < 27:
        age_multiplier = 1.4
    elif age < 30:
        age_multiplier = 1.0
    else:
        age_multiplier = 0.6
    return round(base_value * age_multiplier, 1)

def print_full_report(name, club, age, position,
                      goals, assists, matches,
                      overall, avg_rating):
    # Calculate all values using the return functions
    contributions   = calculate_contributions(goals, assists)
    goals_per_match = calculate_goals_per_match(goals, matches)
    grade, label    = get_performance_grade(avg_rating)
    est_value       = get_transfer_value(overall, age)

    # Build and print the report
    print("=" * 54)
    print(f"  PLAYER RATING REPORT")
    print("=" * 54)
    print(f"  {'Name':<18}: {name}")
    print(f"  {'Club':<18}: {club}")
    print(f"  {'Age':<18}: {age}")
    print(f"  {'Position':<18}: {position}")
    print()
    print(f"  {'SEASON STATISTICS'}")
    print(f"  {'Goals':<18}: {goals}")
    print(f"  {'Assists':<18}: {assists}")
    print(f"  {'Matches':<18}: {matches}")
    print(f"  {'Contributions':<18}: {contributions}")
    print(f"  {'Goals per match':<18}: {goals_per_match:.2f}")
    print()
    print(f"  {'ASSESSMENT'}")
    print(f"  {'Overall rating':<18}: {overall}")
    print(f"  {'Average rating':<18}: {avg_rating:.1f} / 10")
    print(f"  {'Performance grade':<18}: {grade} -- {label}")
    print(f"  {'Est. market value':<18}: £{est_value}M")
    print()
    print("=" * 54)

# Running the engine for three players
print_full_report(
    name       = "Mohamed Salah",
    club       = "Liverpool",
    age        = 31,
    position   = "Right Winger",
    goals      = 22,
    assists    = 13,
    matches    = 35,
    overall    = 89,
    avg_rating = 8.6
)

print_full_report(
    name       = "Dominik Szoboszlai",
    club       = "Liverpool",
    age        = 23,
    position   = "Central Midfielder",
    goals      = 8,
    assists    = 11,
    matches    = 33,
    overall    = 84,
    avg_rating = 7.6
)
```
