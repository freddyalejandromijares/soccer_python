```python
# Interactive goals-per-match calculator
print("=" * 45)
print("   GOALS PER MATCH CALCULATOR")
print("=" * 45)
print()

player_name    = input("Player name: ")
goals          = int(input("Goals scored this season: "))
matches_played = int(input("Matches played: "))

print()

# Guard against dividing by zero
if matches_played > 0:
    goals_per_match = goals / matches_played
    print(f"  {player_name} has scored {goals} goals in {matches_played} matches.")
    print(f"  Goals per match: {goals_per_match:.2f}")
else:
    print("  No matches played yet -- cannot calculate average.")

print()
print("=" * 45)
```

```python
# INTERACTIVE MATCH REPORT BUILDER
# Collects match data from the user and produces a formatted report

print("=" * 50)
print("   MATCH REPORT BUILDER")
print("=" * 50)
print()
print("Please enter the details for today's match.")
print()

# Collect match information
home_team   = input("Home team: ")
away_team   = input("Away team: ")
competition = input("Competition: ")
venue       = input("Venue: ")
attendance  = int(input("Attendance: "))

print()
print("Now enter the final score.")
print()

home_goals = int(input(f"Goals scored by {home_team}: "))
away_goals = int(input(f"Goals scored by {away_team}: "))

# Determine the result
home_win = home_goals > away_goals
away_win = away_goals > home_goals
draw     = home_goals == away_goals

# Build the result label
if home_win:
    result_label = f"{home_team} WIN"
elif away_win:
    result_label = f"{away_team} WIN"
else:
    result_label = "DRAW"

# Display the formatted report
print()
print("=" * 50)
print("   FULL TIME REPORT")
print("=" * 50)
print(f"  {competition}")
print(f"  {venue}  |  Att: {attendance:,}")
print()
print(f"  {home_team:<20} {home_goals}  --  {away_goals}  {away_team}")
print()
print(f"  RESULT: {result_label}")
print("=" * 50)
```
