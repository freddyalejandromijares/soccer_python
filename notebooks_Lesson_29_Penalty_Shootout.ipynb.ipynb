```python
import random

# ── HELPER FUNCTIONS ────────────────────────────────────

def take_penalty(player_name, team_name, score_chance=0.75):
    """
    Simulates one penalty kick.
    Returns True if scored, False if missed.
    score_chance: probability of scoring (default 75%)
    """
    shot = random.random()    # A float between 0.0 and 1.0

    if shot < score_chance:
        print(f"  {player_name} steps up for {team_name}...")
        print(f"  GOAL! The net bulges. {team_name} score!")
        return True
    else:
        print(f"  {player_name} steps up for {team_name}...")
        print(f"  SAVED! The keeper gets down to their right. Miss!")
        return False


def print_score(team_a, score_a, team_b, score_b):
    """
    Displays the current shootout score in a clean format.
    """
    print()
    print(f"  {'SCORE':^44}")
    print(f"  {team_a:^20}  {score_a} -- {score_b}  {team_b:^20}")
    print()


def check_early_winner(score_a, score_b, team_a, team_b,
                       kicks_taken, total_kicks=5):
    """
    Checks if one team cannot be caught with the remaining kicks.
    Returns the winning team name if decided, otherwise returns None.
    """
    remaining = total_kicks - kicks_taken

    # Team A cannot be caught
    if score_a > score_b + remaining:
        return team_a

    # Team B cannot be caught
    if score_b > score_a + remaining:
        return team_b

    # Still alive -- no winner yet
    return None


print("Helper functions defined successfully.")
```

```python
# ── SUDDEN DEATH ────────────────────────────────────────

def run_sudden_death(team_a, squad_a, team_b, squad_b):
    """
    Handles sudden death penalties after a 5-5 tie.
    Continues until one team scores and the other misses in the same round.
    Returns the name of the winning team.
    """
    print()
    print("=" * 52)
    print(f"  {'SUDDEN DEATH':^48}")
    print("=" * 52)
    print()

    round_number = 1

    while True:
        print(f"  -- Sudden Death Round {round_number} --")
        print()

        # Pick a random penalty taker from each squad
        taker_a = random.choice(squad_a)
        taker_b = random.choice(squad_b)

        # Team A kicks first
        scored_a = take_penalty(taker_a, team_a)
        print()

        # Team B kicks
        scored_b = take_penalty(taker_b, team_b)
        print()

        # Check the outcome
        if scored_a and not scored_b:
            print(f"  {team_a} score and {team_b} miss!")
            return team_a

        if scored_b and not scored_a:
            print(f"  {team_b} score and {team_a} miss!")
            return team_b

        if scored_a and scored_b:
            print("  Both teams score. On to the next round!")

        if not scored_a and not scored_b:
            print("  Both teams miss. On to the next round!")

        round_number += 1
        print()


print("Sudden death function defined successfully.")
```

```python
# ── MAIN SHOOTOUT FUNCTION ──────────────────────────────

def run_shootout(team_a, squad_a, team_b, squad_b):
    """
    Runs the five-round penalty shootout.
    Returns the winning team name, or None if sudden death is needed.
    """
    score_a = 0
    score_b = 0

    print()
    print("=" * 52)
    print(f"  {'PENALTY SHOOTOUT BEGINS':^48}")
    print("=" * 52)
    print()

    for round_num in range(1, 6):

        print(f"  ====  Round {round_num} of 5  ====")
        print()

        # Pick a random penalty taker from each squad
        taker_a = random.choice(squad_a)
        taker_b = random.choice(squad_b)

        # Team A takes their penalty
        if take_penalty(taker_a, team_a):
            score_a += 1

        print()

        # Team B takes their penalty
        if take_penalty(taker_b, team_b):
            score_b += 1

        print()

        # Display the score after each round
        print_score(team_a, score_a, team_b, score_b)

        # Check for an early winner after each round
        early_winner = check_early_winner(
            score_a, score_b,
            team_a, team_b,
            kicks_taken=round_num
        )

        if early_winner:
            print(f"  {early_winner.upper()} WIN THE SHOOTOUT EARLY!")
            print(f"  The remaining kicks are not needed.")
            return early_winner

    # After five rounds -- check the result
    if score_a > score_b:
        return team_a
    elif score_b > score_a:
        return team_b
    else:
        return None    # Signal that sudden death is needed


print("Main shootout function defined successfully.")
```

```python
# ── MAIN PROGRAMME ──────────────────────────────────────

def main():
    """
    The entry point. Collects team names and runs the full shootout.
    """
    print("=" * 52)
    print(f"  {'PENALTY SHOOTOUT SIMULATOR':^48}")
    print(f"  {'UEFA Champions League Final':^48}")
    print("=" * 52)
    print()

    # Collect team names
    team_a = input("Enter the name of Team A: ")
    team_b = input("Enter the name of Team B: ")

    # Build the squads
    print()
    print("  Building squads for the shootout...")
    print()

    squad_a = [
        "Alisson", "Alexander-Arnold", "Van Dijk",
        "Robertson", "Mac Allister", "Szoboszlai",
        "Salah", "Nunez", "Diaz", "Jones", "Elliott"
    ]

    squad_b = [
        "Ter Stegen", "Cancelo", "Araujo",
        "Balde", "Pedri", "Gavi",
        "Yamal", "Lewandowski", "Raphinha",
        "Ferran Torres", "De Jong"
    ]

    # Shuffle the squads so takers come in a random order
    random.shuffle(squad_a)
    random.shuffle(squad_b)

    # Run the main five-round shootout
    winner = run_shootout(team_a, squad_a, team_b, squad_b)

    # Handle sudden death if needed
    if winner is None:
        print("  After five rounds -- the scores are level!")
        print("  We go to sudden death...")
        winner = run_sudden_death(team_a, squad_a,
                                  team_b, squad_b)

    # Announce the winner
    print()
    print("=" * 52)
    print(f"  {'SHOOTOUT COMPLETE':^48}")
    print("=" * 52)
    print()
    print(f"  WINNER: {winner.upper()}")
    print()
    print(f"  {winner} win the Champions League!")
    print(f"  The celebrations begin at the Estadio da Luz!")
    print()
    print("=" * 52)


# Run the programme
main()
```

```python
# Run 100 simulations and count wins
wins_a = 0
wins_b = 0

for _ in range(100):
    winner = run_shootout("Liverpool", squad_a,
                          "Barcelona", squad_b)
    if winner == "Liverpool":
        wins_a += 1
    elif winner == "Barcelona":
        wins_b += 1

print(f"Liverpool wins: {wins_a}")
print(f"Barcelona wins: {wins_b}")
```
