```python
# Squad availability calculator — pre-match check
# Used before every fixture to confirm how many players the manager can select

squad = 25          # Total registered first-team players
injured = 3         # Players currently ruled out with injuries
suspended = 1       # Players serving a suspension this match

# Calculate how many players are actually available for selection
available = squad - injured - suspended

print("Available players:", available)
print("Squad size:", squad)
```
