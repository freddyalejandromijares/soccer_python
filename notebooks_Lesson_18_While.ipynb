```python
# Warm-up lap counter
lap = 1

while lap <= 5:
    print(f"Lap {lap} complete. Keep going!")
    lap = lap + 1

print("Warm-up finished. Time for the team talk!")
```

```python
# Using += to update the loop variable
goals_needed = 3
goals_scored = 0

while goals_scored < goals_needed:
    goals_scored += 1    # Same as: goals_scored = goals_scored + 1
    print(f"GOAL! The score is now {goals_scored}.")

print(f"Hat-trick complete! {goals_scored} goals scored.")
```

```python
# Countdown to kick-off
minutes_to_kickoff = 5

print("KICK-OFF COUNTDOWN:")
print()

while minutes_to_kickoff > 0:
    print(f"  {minutes_to_kickoff} minute(s) to kick-off...")
    minutes_to_kickoff -= 1

print()
print("KICK-OFF! The match has begun!")
```

```python
# Make sure the user enters a valid number of goals
# Goals cannot be negative and should not exceed 20 in a single match

print("=== GOAL ENTRY VALIDATOR ===")
print()

goals = int(input("Enter goals scored (0-20): "))

while goals < 0 or goals > 20:
    print("Invalid entry. Goals must be between 0 and 20.")
    goals = int(input("Please try again (0-20): "))

print()
print(f"Valid entry recorded: {goals} goals.")
```

```python
# Season goals accumulator
# Adds up goals from multiple matches until the user is done

print("=== SEASON GOALS TRACKER ===")
print()
print("Enter the goals scored in each match.")
print("Type -1 when you are finished.")
print()

total_goals  = 0
matches      = 0
goals        = 0

while goals != -1:
    goals = int(input(f"Match {matches + 1} goals (-1 to finish): "))
    if goals != -1:
        total_goals += goals
        matches     += 1

print()
print("=" * 38)
print(f"  Matches entered:  {matches}")
print(f"  Total goals:      {total_goals}")

if matches > 0:
    average = total_goals / matches
    print(f"  Goals per match:  {average:.2f}")

print("=" * 38)
```

```python
# MINUTE-BY-MINUTE MATCH SIMULATOR

print("=" * 48)
print(f"{'MATCH SIMULATOR':^48}")
print("=" * 48)
print()

home_team   = input("Home team: ")
away_team   = input("Away team: ")

print()
print(f"  {home_team} vs {away_team}")
print("  Kick-off!")
print()

home_goals = 0
away_goals = 0
minute     = 1

while minute <= 90:

    # A goal is scored every 20 minutes for demonstration purposes
    if minute == 23:
        home_goals += 1
        print(f"  {minute}' -- GOAL! {home_team} score!")
        print(f"  {home_team} {home_goals} -- {away_goals} {away_team}")
        print()

    elif minute == 41:
        away_goals += 1
        print(f"  {minute}' -- GOAL! {away_team} equalise!")
        print(f"  {home_team} {home_goals} -- {away_goals} {away_team}")
        print()

    elif minute == 45:
        print(f"  HALF TIME: {home_team} {home_goals} -- {away_goals} {away_team}")
        print()

    elif minute == 67:
        home_goals += 1
        print(f"  {minute}' -- GOAL! {home_team} retake the lead!")
        print(f"  {home_team} {home_goals} -- {away_goals} {away_team}")
        print()

    elif minute == 90:
        print(f"  FULL TIME: {home_team} {home_goals} -- {away_goals} {away_team}")
        print()

    minute += 1

# Final result
print("=" * 48)
if home_goals > away_goals:
    print(f"  RESULT: {home_team} WIN!")
elif home_goals == away_goals:
    print("  RESULT: DRAW!")
else:
    print(f"  RESULT: {away_team} WIN!")
print("=" * 48)
```
