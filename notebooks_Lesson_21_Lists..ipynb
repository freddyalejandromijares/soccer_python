```python
# DYNAMIC SQUAD MANAGER
# Manage a club's squad using a live list

print("=" * 50)
print(f"{'SQUAD MANAGER':^50}")
print("=" * 50)
print()

club_name = input("Enter your club name: ")
print()

# Starting squad
squad = [
    "Ederson",
    "Walker",
    "Dias",
    "Gvardiol",
    "De Bruyne",
    "Rodri",
    "Foden",
    "Doku",
    "Haaland",
]

print(f"  {club_name.upper()} -- CURRENT SQUAD")
print("-" * 50)

for i in range(len(squad)):
    print(f"  {i + 1:>2}. {squad[i]}")

print("-" * 50)
print(f"  Total players: {len(squad)}")
print()

# New signing
new_signing = input("Sign a new player (enter name): ")
squad.append(new_signing)
print(f"  {new_signing} has signed for {club_name}!")
print()

# Player departure
departing = input("Which player is leaving? (enter name): ")

if departing in squad:
    squad.remove(departing)
    print(f"  {departing} has left {club_name}.")
else:
    print(f"  {departing} is not in the squad. No changes made.")

print()

# Sort the squad alphabetically
squad.sort()

# Final squad report
print("=" * 50)
print(f"  {club_name.upper()} -- UPDATED SQUAD (A-Z)")
print("=" * 50)

for i in range(len(squad)):
    print(f"  {i + 1:>2}. {squad[i]}")

print("-" * 50)
print(f"  Total players: {len(squad)}")
print(f"  First on list: {squad[0]}")
print(f"  Last on list:  {squad[-1]}")
print("=" * 50)
```
