```python
goals_per_match = 27 / 32     # Raw result: 0.84375

# Round to different numbers of decimal places
print(round(goals_per_match, 2))    # 0.84
print(round(goals_per_match, 1))    # 0.8
print(round(goals_per_match, 0))    # 1.0
```

```python
# TRANSFER WINDOW BUDGET CALCULATOR
# Uses integers and floats to manage a club's transfer activity

print("=" * 50)
print(f"{'TRANSFER WINDOW CALCULATOR':^50}")
print("=" * 50)
print()

# Club financial data - integers for counts, floats for money
total_budget     = 200        # int - total budget in millions
season_wage_bill = 3.2        # float - weekly wage bill in millions

# Incoming transfers - floats for fees, ints for wage increase
player_1_fee     = 87.5       # Signing fee
player_1_wage    = 0.35       # Weekly wage increase in millions
player_2_fee     = 45.0
player_2_wage    = 0.22
player_3_fee     = 18.75
player_3_wage    = 0.12

# Outgoing transfer - raises money
sale_fee         = 32.0

# Calculations
total_spent      = player_1_fee + player_2_fee + player_3_fee
net_spend        = total_spent - sale_fee
remaining_budget = total_budget - net_spend
new_wage_bill    = season_wage_bill + player_1_wage + player_2_wage + player_3_wage

# Squad count
original_squad   = 25         # int
players_signed   = 3          # int
players_sold     = 1          # int
new_squad_size   = original_squad + players_signed - players_sold

# Display the report
print(f"  {'BUDGET SUMMARY'}")
print(f"  {'Starting budget':<28}: £{total_budget}M")
print(f"  {'Total spent on signings':<28}: £{total_spent:.1f}M")
print(f"  {'Raised from sales':<28}: £{sale_fee:.1f}M")
print(f"  {'Net spend':<28}: £{net_spend:.1f}M")
print(f"  {'Remaining budget':<28}: £{remaining_budget:.1f}M")
print()
print(f"  {'WAGE BILL'}")
print(f"  {'Original weekly bill':<28}: £{season_wage_bill:.2f}M")
print(f"  {'New weekly bill':<28}: £{new_wage_bill:.2f}M")
print(f"  {'Weekly increase':<28}: £{new_wage_bill - season_wage_bill:.2f}M")
print()
print(f"  {'SQUAD SIZE'}")
print(f"  {'Original squad':<28}: {original_squad} players")
print(f"  {'Signed':<28}: {players_signed} players")
print(f"  {'Sold':<28}: {players_sold} player")
print(f"  {'New squad size':<28}: {new_squad_size} players")
print()

# Final assessment
if remaining_budget >= 50:
    budget_status = "Healthy -- Room for more signings"
elif remaining_budget >= 20:
    budget_status = "Tight -- Limited further activity"
else:
    budget_status = "Stretched -- No major signings possible"

print(f"  {'Budget status':<28}: {budget_status}")
print()
print("=" * 50)
```
