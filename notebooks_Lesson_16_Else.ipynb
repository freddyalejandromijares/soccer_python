```python
# Player fitness check
print("=== FITNESS CHECKER ===")
print()

player_name    = input("Player name: ")
fitness_score  = int(input("Fitness score (0-100): "))

print()

if fitness_score >= 75:
    print(f"  {player_name} is FIT to play.")
    print(f"  Fitness score of {fitness_score} clears the minimum threshold.")
    print("  The player is available for selection.")
else:
    print(f"  {player_name} is NOT fit to play.")
    print(f"  Fitness score of {fitness_score} is below the required 75.")
    print("  The player will not be considered for this fixture.")
```

```python
# Match result classifier using nested if/else
goals_scored   = 3
goals_conceded = 1

if goals_scored > goals_conceded:
    # We won -- now check by how many
    margin = goals_scored - goals_conceded
    if margin >= 3:
        print("DOMINANT WIN! Outstanding performance.")
    else:
        print("WIN! Solid result for the team.")
else:
    # We did not win -- was it a draw or a loss?
    if goals_scored == goals_conceded:
        print("DRAW. One point on the board.")
    else:
        print("DEFEAT. Back to the training ground.")
```

```python
# MATCH RESULT ANALYSER
# Produces a tailored post-match report based on the result

print("=" * 52)
print(f"{'POST-MATCH RESULT ANALYSER':^52}")
print("=" * 52)
print()

# Collect match data
home_team      = input("Home team: ")
away_team      = input("Away team: ")
home_goals     = int(input(f"Goals for {home_team}: "))
away_goals     = int(input(f"Goals for {away_team}: "))
attendance     = int(input("Attendance: "))
shots_on_target = int(input("Shots on target (home team): "))
possession     = float(input(f"Possession % for {home_team}: "))

# Calculated values
goal_difference = home_goals - away_goals
away_possession = 100.0 - possession

print()
print("=" * 52)
print(f"  {home_team}  {home_goals} -- {away_goals}  {away_team}")
print(f"  Attendance: {attendance:,}")
print("=" * 52)
print()

# Main result analysis
if home_goals > away_goals:
    result_label = "HOME WIN"
    points_earned = 3
    print(f"  RESULT: {result_label}")
    print(f"  {home_team} take all three points tonight.")
    print(f"  A strong performance in front of {attendance:,} fans.")
else:
    result_label = "HOME DID NOT WIN"
    points_earned = 0
    print(f"  RESULT: {result_label}")
    print(f"  {home_team} will be disappointed with this outcome.")
    print("  The manager faces tough questions after the final whistle.")

print()

# Possession analysis
print("-" * 52)
print("  POSSESSION ANALYSIS")
print("-" * 52)

if possession >= 55.0:
    print(f"  {home_team} dominated with {possession:.1f}% of the ball.")
    print("  Strong territorial control throughout the match.")
else:
    print(f"  {home_team} had only {possession:.1f}% possession.")
    print(f"  {away_team} controlled the tempo for large periods.")

print()

# Attacking threat analysis
print("-" * 52)
print("  ATTACKING ANALYSIS")
print("-" * 52)

if shots_on_target >= 6:
    print(f"  Excellent: {shots_on_target} shots on target.")
    print("  The attack created plenty of clear-cut chances.")
else:
    print(f"  Limited: only {shots_on_target} shots on target.")
    print("  The attack struggled to test the opposition keeper.")

print()
print("=" * 52)
print(f"  Points earned tonight: {points_earned}")
print("=" * 52)
```
