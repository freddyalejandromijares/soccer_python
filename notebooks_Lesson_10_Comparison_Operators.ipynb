```python
# LEAGUE TABLE ANALYSER
# Evaluates a club's current position and form

# Club data
club_name       = "Aston Villa"
matches_played  = 32
wins            = 19
draws           = 6
losses          = 7
goals_scored    = 74
goals_conceded  = 42
points          = (wins * 3) + (draws * 1)
goal_difference = goals_scored - goals_conceded
goals_per_match = goals_scored / matches_played

# Threshold values
title_race_pts      = 68
champions_league_pts = 60
europa_league_pts    = 52
relegation_risk_pts  = 35
prime_age_start      = 24
prime_age_end        = 30

# Status comparisons
in_title_race       = points >= title_race_pts
champions_league    = points >= champions_league_pts
europa_league       = points >= europa_league_pts
safe_from_rel       = points > relegation_risk_pts
positive_gd         = goal_difference > 0
strong_attack       = goals_per_match >= 2.0
winning_record      = wins > losses

# OUTPUT
print("=" * 50)
print("   LEAGUE STANDING REPORT")
print("=" * 50)
print(f"  Club:            {club_name}")
print(f"  Played:          {matches_played}")
print(f"  Record:          W{wins} D{draws} L{losses}")
print(f"  Points:          {points}")
print(f"  Goal difference: {goal_difference:+}")
print(f"  Goals per match: {goals_per_match:.2f}")
print()
print("-" * 50)
print("  STATUS CHECKS")
print("-" * 50)
print(f"  Title race:          {in_title_race}")
print(f"  Champions League:    {champions_league}")
print(f"  Europa League:       {europa_league}")
print(f"  Safe from rel.:      {safe_from_rel}")
print(f"  Positive GD:         {positive_gd}")
print(f"  Strong attack:       {strong_attack}")
print(f"  Winning record:      {winning_record}")
print()
print("=" * 50)
```
