```python
player = {
    "name":     "Erling Haaland",
    "age":      24,
    "club":     "Manchester City",
    "goals":    27,
    "assists":  5,
    "matches":  29
}

# Accessing values by key
print(f"Name:     {player['name']}")
print(f"Age:      {player['age']}")
print(f"Club:     {player['club']}")
print(f"Goals:    {player['goals']}")
```

```python
# Using dictionary values in calculations
contributions   = player["goals"] + player["assists"]
goals_per_match = player["goals"] / player["matches"]

print(f"Contributions:   {contributions}")
print(f"Goals per match: {goals_per_match:.2f}")
```

```python
# PLAYER SCOUTING PROFILE GENERATOR
# Collects player data and builds a complete dictionary profile

print("=" * 54)
print(f"{'PLAYER SCOUTING PROFILE':^54}")
print("=" * 54)
print()

# Collect player information from the user
name        = input("Player name: ")
age         = int(input("Age: "))
nationality = input("Nationality: ")
club        = input("Current club: ")
position    = input("Position: ")
goals       = int(input("Goals this season: "))
assists     = int(input("Assists this season: "))
matches     = int(input("Matches played: "))
market_val  = float(input("Market value (millions): "))
rating      = float(input("Average match rating (1-10): "))

# Build the player dictionary
player = {
    "name":         name,
    "age":          age,
    "nationality":  nationality,
    "club":         club,
    "position":     position,
    "goals":        goals,
    "assists":      assists,
    "matches":      matches,
    "market_value": market_val,
    "rating":       rating
}

# Add calculated fields to the dictionary
player["contributions"]   = player["goals"] + player["assists"]
player["goals_per_match"] = round(player["goals"] / player["matches"], 2)

# Assign a scouting verdict based on rating
if player["rating"] >= 8.0:
    player["verdict"] = "World Class -- Immediate priority signing"
elif player["rating"] >= 7.0:
    player["verdict"] = "Excellent -- Strong recommendation"
elif player["rating"] >= 6.0:
    player["verdict"] = "Good -- Worth monitoring closely"
else:
    player["verdict"] = "Below standard -- Not recommended at this time"

# Display the complete scouting report
print()
print("=" * 54)
print(f"  SCOUTING REPORT: {player['name'].upper()}")
print("=" * 54)
print()
print(f"  {'PERSONAL INFORMATION'}")
print(f"  {'Age':<20}: {player['age']}")
print(f"  {'Nationality':<20}: {player['nationality']}")
print(f"  {'Current club':<20}: {player['club']}")
print(f"  {'Position':<20}: {player['position']}")
print()
print(f"  {'SEASON STATISTICS'}")
print(f"  {'Goals':<20}: {player['goals']}")
print(f"  {'Assists':<20}: {player['assists']}")
print(f"  {'Matches played':<20}: {player['matches']}")
print(f"  {'Contributions':<20}: {player['contributions']}")
print(f"  {'Goals per match':<20}: {player['goals_per_match']:.2f}")
print()
print(f"  {'FINANCIAL'}")
print(f"  {'Market value':<20}: £{player['market_value']:,.1f}M")
print(f"  {'Match rating':<20}: {player['rating']:.1f} / 10")
print()
print("-" * 54)
print(f"  VERDICT: {player['verdict']}")
print("-" * 54)

# Check for standout qualities
print()
print("  STANDOUT FLAGS:")
found_flag = False

if player["goals"] >= 20:
    print(f"  ✓ Elite scorer -- {player['goals']} goals this season")
    found_flag = True

if player["assists"] >= 10:
    print(f"  ✓ Creative force -- {player['assists']} assists this season")
    found_flag = True

if player["age"] <= 22:
    print(f"  ✓ Young talent -- only {player['age']} years old")
    found_flag = True

if player["market_value"] <= 30:
    print(f"  ✓ Value signing -- available for £{player['market_value']}M")
    found_flag = True

if not found_flag:
    print("  No standout flags raised.")

print()
print("=" * 54)
```
