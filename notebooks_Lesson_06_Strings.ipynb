```python
player_name   = "Vinicius Junior"        # Double quotes
club          = 'Real Madrid'            # Single quotes
match_report  = """Full time.
Real Madrid win the Champions League."""  # Triple quotes for multi-line text
print(player_name)
print(club)
print(match_report)
```

```python
# Building a match announcement from separate parts
home_team  = "Arsenal"
away_team  = "Chelsea"
competition = "Premier League"

announcement = home_team + " vs " + away_team
print(announcement)

full_announcement = competition + ": " + home_team + " vs " + away_team
print(full_announcement)
```
